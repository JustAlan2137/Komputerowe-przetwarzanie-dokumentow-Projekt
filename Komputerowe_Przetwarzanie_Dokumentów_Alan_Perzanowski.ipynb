```python
#1.1. Character encoding: ASCII, Unicode, and UTF-8
#  1.Find Unicode code points.
characters = ["A", "a", "7", "ą", "€"]

for character in characters:
    print(character, ord(character))

# 2.Convert a code point into a character.
print(chr(65))

# 3.Encode English and Polish words as UTF-8 bytes.
print("hello".encode("utf-8"))
print("żółw".encode("utf-8"))

# 4.Explain why ASCII cannot represent every character used in Polish.
#ASCII has only 128 character codes and does not include Polish letters such as ą, ę, or ł. Unicode includes these characters. len("ą") returns 1, whereas len("ą".encode("utf-8")) returns 2: this character takes two bytes in UTF-8.
```

```python
#2.6. Remove stop words
text = "Readability is a crucial quality for any written text, determining how easily a reader can comprehend words, phrases, and overall concepts. Authors often ask themselves: Is this content engaging and accessible to the target audience? By keeping sentences concise, utilizing formatting like bullet points, and avoiding excessive use of jargon or complex abbreviations (such as e.g. or misc.), writers can significantly improve the flow of their work."

tokens = word_tokenize(text.lower())
stop_words = set(stopwords.words("english"))

filtered_words = [
    word for word in tokens
    if word.isalpha() and word not in stop_words
]

print(filtered_words)
```

```python
#2.8. Most common words: FreqDist and Counter
text = (
    "Readability is a crucial quality for any written text, determining how easily a reader can comprehend words, phrases, and overall concepts. Authors often ask themselves: Is this content engaging and accessible to the target audience? By keeping sentences concise, utilizing formatting like bullet points, and avoiding excessive use of jargon or complex abbreviations (such as e.g. or misc.), writers can significantly improve the flow of their work."
)

tokens = word_tokenize(text.lower())
words = [word for word in tokens if word.isalpha()]

nltk_counts = FreqDist(words)
counter_counts = Counter(words)

print("FreqDist:", nltk_counts.most_common(10))
print("Counter:", counter_counts.most_common(10))

print("Same counts:", dict(nltk_counts) == dict(counter_counts))
print(
    "Same ranking:",
    nltk_counts.most_common(10) == counter_counts.most_common(10)
)
```

```python
#3.2. Stemming a Text
text = (
"Readability is a crucial quality for any written text, determining how easily a reader can comprehend words, phrases, and overall concepts. Authors often ask themselves: Is this content engaging and accessible to the target audience? By keeping sentences concise, utilizing formatting like bullet points, and avoiding excessive use of jargon or complex abbreviations (such as e.g. or misc.), writers can significantly improve the flow of their work."
)

stemmer = PorterStemmer()
tokens = word_tokenize(text.lower())

for word in tokens:
    if word.isalpha():
        print(word, "->", stemmer.stem(word))

# Stems are not always English words. For example, the stemmer changes "quality" to "qualiti".
```

```python
#3.3. Lemmatization and parts of speech
lemmatizer = WordNetLemmatizer()

# Each word is paired with its intended part of speech.
words = [
    ("cats", "n"),
    ("running", "v"),
    ("studies", "v"),
    ("better", "a"),
    ("went", "v"),
    ("children", "n")
]

for word, pos in words:
    default_lemma = lemmatizer.lemmatize(word)
    specified_lemma = lemmatizer.lemmatize(word, pos=pos)

    print(
        word,
        "| Default:", default_lemma,
        "| With POS:", specified_lemma
    )
```

```python
#Part 4 Combined Practical Assignment
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk import FreqDist
from nltk.stem import PorterStemmer

# Download the required resources.
nltk.download("punkt_tab")
nltk.download("stopwords")

# 1. Read text from a variable.
text = (
"Readability is a crucial quality for any written text, determining how easily a reader can comprehend words, phrases, and overall concepts. Authors often ask themselves: Is this content engaging and accessible to the target audience? By keeping sentences concise, utilizing formatting like bullet points, and avoiding excessive use of jargon or complex abbreviations (such as e.g. or misc.), writers can significantly improve the flow of their work."
)

# 2. Convert the text to lowercase.
lowercase_text = text.lower()

# 3. Split the text into sentences and tokens.
sentences = sent_tokenize(lowercase_text)
tokens = word_tokenize(lowercase_text)

# 4. Remove punctuation by keeping alphabetic tokens.
words = [token for token in tokens if token.isalpha()]

# 5. Remove English stop words.
stop_words = set(stopwords.words("english"))

filtered_words = [
    word for word in words
    if word not in stop_words
]

# 6. Display the sentence count and remaining word count.
print("Number of sentences:", len(sentences))
print("Number of remaining words:", len(filtered_words))

# 7. Display the 10 most common words.
frequencies = FreqDist(filtered_words)

print("\n10 most common words:")

for word, count in frequencies.most_common(10):
    print(word, ":", count)

# 8. Apply stemming and compare with the original words.
stemmer = PorterStemmer()

print("\nOriginal word -> Stem:")

for word in filtered_words:
    print(word, "->", stemmer.stem(word))
```
